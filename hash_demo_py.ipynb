```python
import hashlib

# Input data
message = "password123"

# Generate SHA-256 hash
hash_value = hashlib.sha256(
    message.encode("utf-8")
).hexdigest()

# Display results
print("Original:", message)
print("SHA-256:", hash_value)
print("Length:", len(hash_value))
```
